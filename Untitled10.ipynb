```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

area = np.array([200, 300, 500, 600, 1000, 1200])
price = np.array([30, 40, 55, 70, 80, 95])

plt.scatter(area, price)
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()


# Method 1: Using first two points

slope = (price[0] - price[1]) / (area[0] - area[1])
print("slope:", slope)

intercept = price[0] - (slope * area[0])
print("intercept:", intercept)

price_prediction = (slope * area) + intercept
print(price_prediction)

plt.figure()
plt.scatter(area, price)
plt.plot(area, price_prediction)
plt.show()


# Method 2: Calculating slope using mean (Least Squares)

area_mean = np.mean(area)
price_mean = np.mean(price)

print(area_mean)
print(price_mean)

num = np.sum((area - area_mean) * (price - price_mean))
print(num)

den = np.sum(np.square(area - area_mean))
print(den)

m = num / den
print(m)

c = price_mean - (m * area_mean)
print(c)

price_pred = (m * area) + c
print(price_pred)

plt.figure()
plt.scatter(area, price)
plt.plot(area, price_pred)
plt.show()


# Method 3: Using Scikit-Learn

from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(area.reshape(-1, 1), price)

y_pred = lr.predict(area.reshape(-1, 1))

plt.figure()
plt.scatter(area, price)
plt.plot(area, y_pred)
plt.show()
```
